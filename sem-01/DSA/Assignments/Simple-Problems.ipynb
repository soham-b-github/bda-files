Q1. Define a procedure that prints histogram() which takes in a list of integers and prints a histogram to the screen. For example, it should print the following if histogram([4,9,7])
****
*********
*******

```python
def histogram(L):
  for i in range(len(L)):
    print(L[i]*'*')


L = list(map(int, input().split()))
print('Histogram is the following')
histogram(L)
```

Q2. Count the number of alphanumeric characters and the number of words

```python
s = input()

alphanum = 0
words = 1

for i in s:
  if i != ' ':
    alphanum += 1
  else:
    words += 1

print('Number of alphanumeric characters = ',alphanum)
print('Number of words = ',words)
```

Q3. WAP to compute the net amount in a bank based on last few transaction details entered by the user

```python
n = int(input('Enter the number of transaction details : '))
print('Enter D for Deposit and W for withdrawal and write the amount involved in the transaction')

net = 0
for _ in range(n):
  trans = input(f'Enter the {_+1}th transaction : ')
  type_trans = trans.split()[0]
  amount = trans.split()[1]
  if type_trans == 'D':
    net += int(amount)
  else:
    net -= int(amount)
  if(net<=0):
    print('Bankrupt')
    break


print(f'Net balance in Bank : Rs {net}')
```

Q4. Read and process the same procudure using a file

```python
path = 'transactions.txt'

n = int(input('Enter the number of transaction details : '))
print('Enter D for Deposit and W for withdrawal and write the amount involved in the transaction')

net = 0

with open(path, 'w+') as f:
  f.write('Operation\tAmount_of_transaction\tBank_Balance\n')
  for _ in range(n):
    trans = input(f'Enter the {_+1}th transaction : ')

    line = ''
    type_trans = trans.split()[0]
    amount = trans.split()[1]

    line = type_trans+'\t\t\t'+amount+'\t\t'

    if type_trans == 'D':
      net += int(amount)
    else:
      net -= int(amount)

    if(net<=0):
      line+='Bankrupt'
      f.write(line)
      # print('Bankrupt')
      break
    else:
      line+=str(net)+'\n'
      f.write(line)


print('Bank Balance = Rs. ',net)

with open(path, 'r') as f:
  print(f.read())
```

Q5. Check a number to be prime

```python
def isprime(n):
  for i in range(2,n):
    if n%i==0:
      return 'is Prime'
  return 'is Prime'


n = int(input('Enter a number = '))
print(f'{n} {isprime(n)}')
```

Q6. Define a function which can generate a list of squares of numbers ranging from 1 to 10. And then print only the last 5 values of the generated list

```python
def generate_list():
  L = [i**2 for i in range(1,11)]
  return L


L = generate_list()
print(L[-5:])
```

Q7. WAP to print all possible combinations of subject, verb, object where subject = [I, You] and Object = [Football, Hockey] and Verb = [Play, Love]

```python
subject = ['I','You']
verb = ['Play','Love']
object = ['Football','Hockey']

lines = ''

for i in subject:
  for j in verb:
    for k in object:
      lines+=f'{i} {j} {k}\n'

path = 'sentences-assignment02.txt'

with open(path, 'w+') as f:
  f.write(lines)

with open(path, 'r') as f:
  print(f.read())
```
