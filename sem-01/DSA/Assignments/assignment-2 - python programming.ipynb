**Question No. 1**
Find the maximum of two numbers without using any predefined functions

```python
def maximum(a,b):
  return a if a>b else b


a = int(input('Enter first number = '))
b = int(input('Enter second number = '))
print(f'Maximum of {a} and {b} is {maximum(a,b)}')
```

**Question No. 2**
FInd maximum of three numbers using user-defined functions

```python
def max_of_three(a,b,c):
  return a if a>b and a>c else b if b>c else c


a = int(input('Enter first number = '))
b = int(input('Enter second number = '))
c = int(input('Enter third number = '))

print(f'Maximum of {a},{b} and {c} is {max_of_three(a,b,c)}')
```

**Question No. 3** Find the length of a string using user defined functions

**Question No. 5**
WAP to define a function named 'translate' that takes text as an argument and doubles every consonant and places an 'o' in between

```python
def translate(text):
  vowels = 'AEIOUaeiou'
  translated_text = ''
  for char in text:
    if char not in vowels:
      translated_text += char + 'o' + char
    else:
      translated_text += char

  return translated_text


text = input('Enter a text = ')
print(f'Translated text = {translate(text)}')
```

**Question No. 6**
Define a function sum() and multiply() that sums and multiplies all the numbers of the list and returns the answer

```python
def sum(l1):
  s = 0
  for i in l1:
    s+=i
  return s


def multiply(l1):
  m = 1
  for i in l1:
    m*=i
  return m


l1 = [int(i) for i in input('Enter numbers (give spaces in between)= ').split(' ')]
print(l1)
print(f'Sum = {sum(l1)}')
print(f'Multiplication = {multiply(l1)}')
```

**Question No. 8** Palindrome Check

```python
def is_palindrome(string):
  reverse = string[::-1]
  return True if reverse==string else False


text = input('Enter a string : ')
print(f'Is {text} a palindrome? {is_palindrome(text.lower())}')
```

**Question No. 9** Check whether an element is a part of a list using a user-defined function. Both list and the element to be checked are user-inputs.

```python
def is_member(a, l):
  for i in l:
    if i==a:
      return "Yes"
  return "No"


l = [i for i in input('Enter list of elements (add spaces in between elements) : ').split(' ')]
a = input('Enter element to be checked : ')
print(f'Is {a} a member of {l}? {is_member(a,l)}')
```

**Question No. 10**
Checks whether two lists have common elements in them

```python
def common_elements(l1,l2):
  for i in l1:
    if i in l2:
      return "Yes"
  return "No"


l1 = [i for i in input('Enter list 1 of elements (add spaces in between elements) : ').split(' ')]
l2 = [i for i in input('Enter list 2 of elements (add spaces in between elements) : ').split(' ')]
print(f'Are there common elements in {l1} and {l2}? {common_elements(l1,l2)}')
```
