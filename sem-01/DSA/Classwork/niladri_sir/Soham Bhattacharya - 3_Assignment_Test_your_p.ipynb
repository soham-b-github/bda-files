## Problem 2: Sum of Digits
Task: Write a Python function to calculate the sum of digits of a given number. Hint: Convert the number to a string, iterate over each character, and sum the integers.

Help: w3schools for numbers, how to iterate on iterables. What are iterables etc.

````python
def sum_of_digits(num):
    # write your code here
    s=0
    while num:
      s+=num%10
      num/=10
    return s
````

````python
# write your test cases here and execute them before you submit the assignment file
# Test cases for : sum_of_digits(num)
print(sum_of_digits(10))
print(sum_of_digits(14))
print(sum_of_digits(232))
````

## Problem 9: Key with Maximum Value
Task: Write a Python function to find the key with the maximum value in a dictionary. Hint: Use the max() function with a lambda function. Reference: https://www.w3schools.com/python/python_dictionaries.asp

````python
# write your function
def maxVal(dict):
  keys = list(dict.keys())
  max = dict[keys[0]]
  for key in keys:
    if dict[key]>max:
      max = dict[key]
  return max
````

````python
# test cases
print(maxVal({'x':1,'y':3,'z':5}))
````

## Problem 7: BLACKJACK:

Given three integers between 1 and 11, if their sum is less than or equal to 21, return their sum. If their sum exceeds 21 *and* there's an eleven, reduce the total sum by 10.

Finally, if the sum (even after adjustment) exceeds 21, return 'BUST'

blackjack(5,6,7) --> 18

blackjack(9,9,9) --> 'BUST'

blackjack(9,9,11) --> 19

````python
# write your function
def blackjack(a,b,c):
  L = [a,b,c]
  s = a+b+c
  if s<=21:
    return s
  elif 11 in L:
    return s-10
  else:
    return 'BUST'
````

````python
# test cases
print(blackjack(5,6,7))
print(blackjack(9,9,9))
print(blackjack(9,9,11))
````

## Problem 8: SPY GAME:

Write a function that takes in a list of integers and returns True if it contains 007 in order

spy_game([1,2,4,0,0,7,5]) --> True

spy_game([1,0,2,4,0,5,7]) --> True

spy_game([1,7,2,0,4,5,0]) --> False

````python
# write your function
def spy_game(L):
  zeros = [i for i,x in enumerate(L) if x==0]
  seven = [i for i,x in enumerate(L) if x==7]

  return seven[0]>zeros[-1] and len(zeros)>=2
````

````python
# test cases
print(spy_game([1,2,4,0,0,7,5]))
print(spy_game([1,0,2,4,0,5,7]))
print(spy_game([1,7,2,9,4,5,0]))
````

## Problem 9: SUMMER OF '69:

Return the sum of the numbers in the array, except ignore sections of numbers starting with a 6 and extending to the next 9 (every 6 will be followed by at least one 9). Return 0 for no numbers.

summer_69([1, 3, 5]) --> 9

summer_69([4, 5, 6, 7, 8, 9]) --> 9

summer_69([2, 1, 6, 9, 11]) --> 14

````python
# write your function
def summer_of_69(L):
  flag = False
  sum = 0
  for i in L:
    if i==6:
      flag = True
      continue
    elif i==9:
      flag = False
      continue
    sum+=i if not flag else 0

  return sum
````

````python
# test cases
print(summer_of_69([1,3,5]))
print(summer_of_69([4,5,6,7,8,9]))
print(summer_of_69([2,1,6,9,11]))
````

## Problem 10: COUNT PRIMES:

Write a function that returns the *number* of prime numbers that exist up to and including a given number

count_primes(100) --> 25

By convention, 0 and 1 are not prime.

````python
# write your function
def is_prime(n):
  for i in range(2,int(n**0.5)+1):
    if not n%i:
      return False
  return True

def count_primes(n):
  if n<=2:
    return n==2
  else:
    c=1 # includes number 2 as a prime within range
    for i in range(3,n+1,2):
      c+=1 if is_prime(i) else 0
    return c
````

````python
# test cases
count_primes(100)
````

## Problem 11: UP LOW:

Write a Python function that accepts a string and calculates the number of upper case letters and lower case letters.Sample String : 'Hello Mr. Rogers, how are you this fine Tuesday?'

Expected Output :

No. of Upper case characters : 4

No. of Lower case Characters : 33

HINT: Two string methods that might prove useful:  and
If you feel ambitious, explore the Collections module to solve this problem!


````python
# write your function
def no_case_letters(s):
  lo_case = 'thequickbrownfoxjumpsoverthelazydog'
  up_case = 'THEQUICKBROWNFOXJUMPSOVERTHELAZYDOG'
  lc_count, uc_count = 0,0

  for i in s:
    lc_count+=1 if i in lo_case else 0
    uc_count+=1 if i in up_case else 0

  print(f'No. of Upper case characters : {uc_count}')
  print(f'No. of Lower case characters : {lc_count}')
````

````python
# test cases
no_case_letters('Hello Mr. Rogers, how are you this fine Tuesday?')
````

## Problem 13: palindrome:

Write a Python function that checks whether a word or phrase is palindrome or not.

Note: A palindrome is word, phrase, or sequence that reads the same backward as forward,

e.g., madam,kayak,racecar, or a phrase "nurses run".

Hint: You may want to check out the .replace() method in a string to help out with dealing with spaces. Also google search how to reverse a string in Python, there are some clever ways to do it with slicing notation.

````python
# write your function
def palindrome_check(s):
  L = "".join(s.split())
  return L==L[::-1]
````

````python
# test cases
print(palindrome_check('nurses run'))
print(palindrome_check('madam'))
print(palindrome_check('kayak'))
print(palindrome_check('racecar'))
print(palindrome_check('holiday'))
````

## Problem 14: ispangram:

Write a Python function to check whether a string is pangram or not. (Assume the string passed in does not have any punctuation)

Note : Pangrams are words or sentences containing every letter of the alphabet at least once.
For example : "The quick brown fox jumps over the lazy dog"

Hint: You may want to use .replace() method to get rid of spaces.

Hint: Look at the string module

Hint: In case you want to use set comparisons

````python
# write your function
def ispangram(s):
  L = list(set("".join(s.lower().split())))
  return len(L)==26
````
