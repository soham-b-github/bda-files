# **Requirements:**

**Time :  10 mins**

**Attributes:**

* account_number: A unique identifier for the account.

* balance: The current balance of the account.

**Methods:**

* deposit(amount): Adds the specified amount to the account balance.

* withdraw(amount): Subtracts the specified amount from the account balance. Ensure that the balance doesn't go negative.

* show_balance(): Prints the current account balance.

Write the program in a way, that user continues the baking operation until choice is given to exit.



```python
# write your code

class Bank:

  def __init__(self, account_number):
    self.account_number = account_number
    self.balance = 0

  def deposit(self, amount):
    self.balance+=amount
    print(f"Deposit of {amount} is successful!\nBalance = Rs.{self.balance}")

  def withdraw(self, amount):
    new_bal = self.balance-amount
    if new_bal<0:
      print("Insufficient balance")
    else:
      print(f"Withdrawal of Rs.{amount} is successful!")
      self.balance = new_bal
      print(f"Updated balance = Rs.{self.balance}")

  def show_balance(self):
    print(f"Balance = Rs.{self.balance}")



accnum = int(input('Enter account number of the user :'))
bank = Bank(accnum)

print("\n1. Deposit\n2. Withdraw\n3. Show Balance\n4. Exit")

while True:
  choice = int(input("Enter your choice :"))
  if choice == 1:
    amount = int(input("Enter amount to deposit :"))
    bank.deposit(amount)
  elif choice == 2:
    amount = int(input("Enter amount to withdraw :"))
    bank.withdraw(amount)
  elif choice == 3:
    bank.show_balance()
  elif choice == 4:
    print('Exiting the process!')
    break
```
