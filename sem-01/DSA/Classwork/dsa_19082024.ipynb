## **Function Overloading**

```python
class Point2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Getters
    def get_x(self):
        return self.x

    # Setters
    def set_x(self, x):
        self.x = x

    # Function overloading of built-in addition operation
    def __add__(self, other):
      if(isinstance(other, Point2D)):
        return Point2D(self.x + other.x, self.y + other.y)

      elif(isinstance(other, (tuple, list))):
        return Point2D(self.x + other[0], self.y + other[1])

      elif(isinstance(other, (int, float))):
        return Point2D(other+self.x, other+self.y)

      else:
        return NotImplementedError

    # Function overloading the built-in print method
    def __str__(self):
        return f"(x,y) = ({self.x}, {self.y})"

    # Function overloading the built-in multiplication method
    def __mul__(self, other):
      if(isinstance(other, (int, float))):
        return Point2D(self.x * other, self.y * other)
      elif(isinstance(other, Point2D)):
        return Point2D(self.x*other.x,self.y*other.y)
      else:
        return NotImplementedError

    # Function overloading the built-in rmul method
    def __rmul__(self, other):
      return self.__mul__(other)

    # Function overloading the built-in radd method
    def __radd__(self, other):
      return self.__add__(other)
```

## **Inheritance**



*   Inheriting properties from some superior source
*   For any initialization of instances, we have to use the constructor



```python
class Point3D(Point2D):
  # Here Point2D is the superclass or the parent class
  # Point3D inherits the properties of the Point2D class
  # So Point3D is the child class of Point2D
  def __init__(self, x, y, z=0):
     super().__init__(x, y)
     self.z = z

  def __str__(self):
    return f"(x,y,z) = ({self.x}, {self.y}, {self.z})"
```

## **Doubt Clearance**

```python
# Integers are immutable
# Lists are mutable
def scale(data, scalar):

  # everytime i is updated, i changes it reference to the address of the newly computed value
  # so the value-objects in list 'data' does not change the reference addresses

  for i in data:
    i*=scalar
  return data


def scale1(data, scalar):

  # Since lists are mutable, here, data[i] when assigned to (data[i]*scalar), this new value is an immutable integer object which has a new address reference
  # and in this data[i]*=scalar updation, data[i] changes its reference address from the reference address of the originally stored value to the address reference of
  # the newly computed immutable integer object
  # So the value-objects in list 'data' changes the reference addresses

  for i in range(len(data)):
    data[i]*=scalar
  return data


lst = [1,2,3]
print(scale(lst, 2))
print(scale1(lst, 2))
```
