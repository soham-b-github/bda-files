```python
b = np.arange(24).reshape(6,4)
print(b[1:2,:])
```

### Self-modification

```python
b = np.arange(80).reshape(4,20)
print(b[-1:,:])
```

Basically *np.arange(n)* produces a sequence of integers from 0 to n-1 <br> and *reshape(p,q)* reshapes the sequence into a matrix format (list of lists) : p many lists of q-sized list provided p*q = n

# **Question 1.8**

```python
import pandas as pd
import numpy as np
Salary = pd.DataFrame({
    "bonus" : [300,3111,np.nan,4333,300,5000,np.nan,10000,300,3432,1,900]
})
```

```python
Salary
```

```python
# using the mean as a replacement value for the NaN values

mean_bonus = Salary.bonus.mean()
print('Mean = ',mean_bonus)
Salary_imputedWithMean = Salary.fillna(mean_bonus)
print('New DataFrame:\n',Salary_imputedWithMean)
```

```python
# using the median as a replacement value for the NaN values

median_bonus = Salary.bonus.median()
print('Median = ',median_bonus)
Salary_imputedWithMedian = Salary.fillna(median_bonus)
print('New DataFrame:\n',Salary_imputedWithMedian)
```

```python
# using the mode as a replacement value for the NaN values

mode_bonus = Salary.bonus.mode()
print(f'Number of Modes = {len(mode_bonus)}')
if len(mode_bonus) == 2:
  print(f'Mode = {mode_bonus}')
else:
  print(f'Mode = {mode_bonus[0]}')
Salary_imputedWithMode = Salary.fillna(mode_bonus)
print('New DataFrame:\n',Salary_imputedWithMode)
```
