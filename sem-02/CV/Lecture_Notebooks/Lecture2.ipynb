# What's going on?

## OpenCV uses BGR (Blue-Green-Red) and not RGB (Red-Green-Blue)

https://physiophile.wordpress.com/2017/01/12/why-opencv-uses-bgr-not-rgb/

# matplotlib uses RGB
# <font color='red'> So, Be extreamly careful when using matplot lib to plot images opened by OpenCV </font>

```python
img_what = img1[:,::-1,:]
plt.figure(figsize=(16, 16))
plt.subplot(1, 2, 1)
plt.imshow(img1)

plt.subplot(1, 2, 2)
plt.imshow(img_what)
plt.show()
```

Min and Max value of a channel

```python
img1Blue = img1[:,:,0]
img1Red = img1[:,:,2]



minBlue = img1Blue.min()
maxBlue =img1Blue.max()
meanBlue =img1Blue.mean()
    
minRed = img1Red.min()
maxRed = img1Red.max()
meanRed =img1Red.mean()


print('Min Blue: %d Max Blue: %d Mean Blue: %d' %(minBlue, maxBlue, meanBlue))
print('Min Red: %d Max Red: %d Mean Red: %d' %(minRed, maxRed, meanRed))

```

Add Gaussian noise to green channel


```python
    imgGreen = img1[:,:,1].copy()
    height, width = imgGreen.shape
    
    noise = np.zeros((height,width), np.uint8)
    noise2 = np.zeros((height,width), np.uint8)
    noise3 = np.zeros((height,width), np.uint8)


    # Meaning: cv2.randn(dst, mean, stddev)
    cv2.randn(noise, 0, 30)
    cv2.randn(noise2, 0, 60)
    cv2.randn(noise3, 0, 90)

    imgGreenNoise1 = imgGreen + noise
    imgGreenNoise2 = imgGreen + noise2
    imgGreenNoise3 = imgGreen + noise3

    
    plt.figure(figsize=(32, 32))

    plt.subplot(2, 2, 1)
    plt.imshow(img1[:,:,1],cmap='gray')
    plt.title('ORGINAL')

    plt.subplot(2, 2, 2)
    plt.imshow(imgGreenNoise1,cmap='gray')
    plt.title('SIGMA = 30')

    plt.subplot(2, 2, 3)
    plt.imshow(imgGreenNoise2,cmap='gray')
    plt.title('SIGMA = 60')

    plt.subplot(2, 2, 4)
    plt.imshow(imgGreenNoise3,cmap='gray')
    plt.title('SIGMA = 90')

    plt.show()
   
```
