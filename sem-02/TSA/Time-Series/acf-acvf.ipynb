```python
import numpy as np  # For clearness, imports will only be shown in this snippet.
```

```python
def autocovariance(series, lag, purpose="acvf"):
    len_series = len(series)
    mean_series = np.mean(series)
    if purpose=="acvf":
        print('Mean of the data:',mean_series)
    covariances = []
    for index in range(lag, len_series):
        # Simple covariance calculation. This can be easily replaced with np.cov,
        # but it is educational to calculate it ourselves.
        covariance = (series[index] - mean_series) * (series[index - lag] - mean_series)
        covariances.append(covariance)

    autocovariance = np.sum(covariances) / len_series
    return autocovariance
```

```python
def autocorrelation_function(series, max_lags=20):
    lag_list = list(range(max_lags))
    autocovariances = []
    for lag in lag_list:  # Calculating autocovariances for each lag
        autocovariances.append(autocovariance(series, lag, "acf"))
    
    acvf_0 = autocovariances[0]
    print('ACVF(0) =',acvf_0)
    autocorrelations = np.divide(autocovariances, acvf_0)
    return autocorrelations
```

```python
series = np.array([23.32,32.33,28.98,33.16,26.33,29.88,32.69,18.98,21.23,26.66])
```

```python
autocorrelation_function(series)
```
